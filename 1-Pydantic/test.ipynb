```python
from pydantic import BaseModel

class Person(BaseModel):
    name : str
    age : int 
    city : str


person = Person(name="Jai",age="23",city="delhi")

print(person)
```

```python
person2 = Person(name="Jai",age="23",city=1100) #Validation error coming from pydantic as, datatype of city is not followed
```

```python
## optional fields 

from typing import Optional

class Employee(BaseModel):
    id : int
    name : str 
    department : str
    salary : Optional[float] = None 
    is_active : Optional[bool] = True

emp1 = Employee(id = 123, name="jai", department = "tech")
emp1
```

```python
emp1 = Employee(id = 123, name="jai", department = "tech",salary=1000000)
emp1
```

```python
from typing import List

class Classroom(BaseModel):
    room_number : str 
    students : List[str]
    capacity : int

c1 = Classroom(room_number="A101" , students=["a","b","c"], capacity=12)
c1
```

```python
c2 = Classroom(room_number="A101" , students=("a","b","c"), capacity=12)
c2 # can also typecast internally , as can seen in this example - students tuple got converted into list
```

```python
## catching validation error 

from pydantic import ValidationError

try :
    c3 = Classroom(room_number="A101" , students=("a","b",12), capacity=12)
except ValidationError as e :
    print(f"Invalid value : {e}")
```

```python
## Creating Nested Pydantic Models + Email Validation

from pydantic import EmailStr

class Address(BaseModel): 
    city : str 
    pinCode : int 
    state : str


class Customer(BaseModel):
    id : int 
    name : str 
    email : EmailStr
    address : Address


c1 = Customer(id=1, name = "jai", email="test@gmail.com",address={"city": "new delhi","state": "delhi","pinCode":110096})
c1
```

```python
c2 = Customer(id=1, name = "jai", email="test@gmail.com",address={"city": "new delhi","state": "delhi","pinCode":"110096"})
c2 # here digits str gets converted into int 
```

```python
# Adding constraints in our pydantic model 

from pydantic import Field
class Item(BaseModel) : 
    id : int 
    name : str = Field(min_length=2 , max_length=100)
    price : float = Field(ge=0,le=1000) # greater than & equals to 0 and lower and equals to 1000
    quantity : int = Field(ge=0)

i1 = Item(id = 1 , name = "egg" , price=10 , quantity= 500)
i1
```

```python
i1 = Item(id = 1 , name = "egg" , price=-10 , quantity= 500)
i1 # validation error : price must be greater than 0
```
